```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from xgboost import XGBClassifier
```

```python
df = pd.read_csv("cardio_train.csv")

print("Shape :", df.shape)

df.head()
```

## Basic EDA

```python
df.info()

print("\nMissing Values\n")
print(df.isnull().sum())

print("\nDuplicate Rows :", df.duplicated().sum())

print("\nStatistical Summary\n")
print(df.describe())
```

## Target Distribution

```python
sns.countplot(x="cardio", data=df)

plt.title("Cardiovascular Disease Distribution")
plt.show()
```

## Correlation Heatmap

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```

## Outlier Detection

```python
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
sns.boxplot(y=df["ap_hi"])
plt.title("Systolic BP")

plt.subplot(1,2,2)
sns.boxplot(y=df["ap_lo"])
plt.title("Diastolic BP")

plt.show()
```

## Height & Weight Outliers

```python
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
sns.boxplot(y=df["height"])
plt.title("Height")

plt.subplot(1,2,2)
sns.boxplot(y=df["weight"])
plt.title("Weight")

plt.show()
```

## Data Cleaning

```python
df.drop("id", axis=1, inplace=True)

df = df[
    (df["height"] >= 120) &
    (df["height"] <= 220) &
    (df["weight"] >= 30) &
    (df["weight"] <= 200) &
    (df["ap_hi"] >= 80) &
    (df["ap_hi"] <= 250) &
    (df["ap_lo"] >= 40) &
    (df["ap_lo"] <= 200)
]

print(df.shape)
```

## Feature Engineering

```python
#Age Conversion
df["age"] = df["age"] / 365
```

```python
#BMI
df["bmi"] = df["weight"] / ((df["height"]/100)**2)
```

```python
#Pulse Pressure
df["pulse_pressure"] = df["ap_hi"] - df["ap_lo"]

#Mean Arterial Pressure
df["map"] = (
    2 * df["ap_lo"] + df["ap_hi"]
) / 3
```

```python
# Features & Target
X = df.drop("cardio", axis=1)

y = df["cardio"]

print(X.shape)
print(y.shape)
```

# Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

## Feature Scaling

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

# Model Training

```python
models = {

    "Logistic Regression":
    LogisticRegression(max_iter=2000),

    "Decision Tree":
    DecisionTreeClassifier(max_depth=8),

    "Random Forest":
    RandomForestClassifier(
        n_estimators=300,
        max_depth=10,
        random_state=42
    ),

    "Gradient Boosting":
    GradientBoostingClassifier(
        n_estimators=200,
        learning_rate=0.05,
        random_state=42
    ),

    "XGBoost":
    XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    random_state=42,
    eval_metric="logloss"
),

    "KNN":
    KNeighborsClassifier(7),

    "SVM":
    SVC(
        kernel="rbf",
        
    )
}
```

# Accuracy + ROC-AUC

```python
results = []

for name, model in models.items():

    if name in ["Logistic Regression","KNN","SVM"]:

        model.fit(X_train_scaled,y_train)

        pred = model.predict(X_test_scaled)

        prob = model.predict_proba(X_test_scaled)[:,1]

    else:

        model.fit(X_train,y_train)

        pred = model.predict(X_test)

        prob = model.predict_proba(X_test)[:,1]

    accuracy = accuracy_score(y_test,pred)

    roc_auc = roc_auc_score(y_test,prob)

    results.append(
        [name,accuracy,roc_auc]
    )
```

## Final Comparison Table

```python
result_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "ROC_AUC"
    ]
)

result_df["Accuracy"] *= 100

result_df = result_df.sort_values(
    by="ROC_AUC",
    ascending=False
)

result_df
```

## Best Model

```python
best_model = result_df.iloc[0]

print("Best Model :", best_model["Model"])

print("Accuracy :", round(best_model["Accuracy"],2))

print("ROC-AUC :", round(best_model["ROC_AUC"],4))
```
